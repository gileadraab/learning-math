```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from sympy import symbols, solve, diff, integrate
import math
```

```python
# Week 1 Self-Assessment Quiz
# Try to solve these without looking at your notes first

print("WEEK 1 SELF-ASSESSMENT")
print("======================")

# Question 1: Solve 3(x + 2) = 15
def question1():
  x = sp.symbols('x')
  equation = 3*(x + 2) - 15

  solution = solve(equation, x)
  print("Question 1: Solve 3(x + 2) = 15")
  print("Solution:", solution[0])
  print()

question1()
```

```python
# Question 2: If f(x) = x² + 3x - 1, find f(2)
def question2(y):
  x = sp.symbols('x')
  equation = x**2 + 3*x - 1

  solution = equation.subs(x, y)
  print("Question 2: If f(x) = x² + 3x - 1, find f(2)")
  print("Solution:", solution)
  print()

question2(2)
```

```python
# Question 3: Plot y = x² - 4x + 3 and identify its vertex
x_values = np.linspace(-5, 5, 100)
y_values = x_values**2 - 4*x_values + 3

x = sp.symbols('x')
function = x**2 - 4*x + 3

derivative = sp.diff(function, x)

x_vertex = solve(derivative, x)[0]
y_vertex = function.subs(x, x_vertex)

print("Question 3:")
plt.plot(x_values, y_values)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of y = x² - 4x + 3')
plt.grid(True)
plt.show()

print("Derivative:", derivative)
print("Vertex:", (x_vertex, y_vertex))
print()
```
